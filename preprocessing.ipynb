Label Encoding

```python
#Importing the modules
import pandas as pd
#scikit-learn its an library
#preprocessing module
from sklearn.preprocessing import LabelEncoder,OneHotEncoder
#LabelEncoder is class
```

```python
df=pd.DataFrame({
    'Product':['Watch','Mouse','stove','shoe','stove','Mouse','Watch','shoe'],
    'subjects':['sql','Sql','Testing','genai','Genai','Testing','Testing','genai'],
    'objects':['marker','Duster','Duster','Marker','Pen','Duster','Pencil','Pen']

})
df
```

```python
df['Product'].unique()
```

```python
le=LabelEncoder()
```

```python
#fit will learn the data,identify the unique vales,sort them and map them
#transform will transfrom the data
#.fit() and .transform()
#fit_transform accepts one positional argument
df['Product']=le.fit_transform(df['Product'])
df['subjects']=le.fit_transform(df['subjects'])
df['objects']=le.fit_transform(df['objects'])
df
```

One Hot Encoder

```python
import pandas as pd
#scikit-learn its an library
#preprocessing module
from sklearn.preprocessing import OneHotEncoder
ohe=OneHotEncoder(sparse_output=False)
#OneHotEncoder is class
```

```python
df1=pd.DataFrame({
    'Gadgets':['Watch','Mobile','Mobile','mobile','Tablets','ipad','Watch']

})
df
```

Examples on OneHotEncoder

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
df3=pd.DataFrame({'Batches':['A1','A1','M24','M168','M19','M19','M24']}
                )
ohe=OneHotEncoder(sparse_output=False)
data8=ohe.fit_transform(df3[['Batches']])
name=ohe.get_feature_names_out(['Batches'])
df3=pd.DataFrame(data8,columns=name)
df3
```

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
df2=pd.DataFrame({'Actors':['ALLU ARJUN','PRABHAS','SALMAN KHAN','RANBIR KAPOOR','RANVEER','ALLU ARJUN','PRABHAS']}
                )
ohe=OneHotEncoder(sparse_output=False)
dataa=ohe.fit_transform(df2[['Actors']])
name=ohe.get_feature_names_out(['Actors'])
df2=pd.DataFrame(dataa,columns=name)
df2
```
