```python
#To replacve all the null values with mean
cols=var.select_dtypes(include=['int64','float64']).columns
for i in cols:
  var[i]=var[i].fillna(var[i].mean())
#only int and float have been replaced
```

```python
#To replacve all the null values with mean
cols1=var.select_dtypes(include=['object']).columns
for i in cols1:
  var[i]=var[i].fillna(var[i].mode())
var.isnull().sum()
```

```python
import pandas as pd
var=pd.read_csv("https://github.com/Taj1920/Data-Analysis-Datasets/raw/refs/heads/main/Datasets-Cleaning/medical_students_dataset.csv")
```

```python
#Fill NaN values with the mean (or 0, or median)
var["Age"] = var["Age"].fillna(var["Age"].mean())
#converting dtype of age to int coz age cant be in float
var["Age"] = var["Age"].astype(int)
#to print and check
print(var["Age"].isnull().sum())
```

```python
#Fill NaN values with the mean (or 0, or median)
var["Student ID"] = var["Student ID"].fillna(var["Student ID"].mean())
#converting dtype of student id to int coz student id cant be in float
var["Student ID"] = var["Student ID"].astype(int)#astype converts to praticular datatype which is passed in ()
#to print and check
print(var["Student ID"].isnull().sum())
```

```python
#cleaning the text types
cols=var.select_dtypes(include=['string']).columns
for col in cols:
  var[col]=var[col].str.lower()
#Any irrelevant data or mixed of upper and lower cases
```

```python
#To drop duplicates
var=var.drop_duplicates()
```

```python
#Rename/standarlize the col names
var.columns=var.columns.str.strip()#it removes all the spaces from col-name
var.columns=var.columns.str.lower()#all the col-names shld be in the same format of col-name
```
