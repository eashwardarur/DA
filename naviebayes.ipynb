```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_diabetes
from sklearn.naive_bayes import GaussianNB
d=load_diabetes()
d
```

```python
df=pd.DataFrame(d.data,columns=d.feature_names)
df['target']=d.target
```

```python
out=[]
for i in df.target:
  if i>=120:
    out.append(1)
  else:
    out.append(0)
```

```python
df['diabeties']=out
X=df.drop(['target','diabeties'],axis=1)
y=df.diabeties
```

```python
import pandas as pd
from sklearn.naive_bayes import CategoricalNB
#implements Naive Bayes for categorical features.
from sklearn.preprocessing import LabelEncoder
#converts text values into numbers.

data = {
    'Outlook': ['Sunny','Sunny','Overcast','Rain','Rain','Rain',
                'Overcast','Sunny','Sunny','Rain','Sunny',
                'Overcast','Overcast','Rain'],
    'Temperature': ['Hot','Hot','Hot','Mild','Cool','Cool',
                    'Cool','Mild','Cool','Mild','Mild',
                    'Mild','Hot','Mild'],
    'Humidity': ['High','High','High','High','Normal','Normal',
                 'Normal','High','Normal','Normal','Normal',
                 'High','Normal','High'],
    'Wind': ['Weak','Strong','Weak','Weak','Weak','Strong',
             'Strong','Weak','Weak','Weak','Strong',
             'Strong','Weak','Strong'],
    'Play': ['No','No','Yes','Yes','Yes','No',
             'Yes','No','Yes','Yes','Yes','Yes',
             'Yes','No']
}
df = pd.DataFrame(data)

encoders = {}
for col in df.columns:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])
    encoders[col] = le

X = df.drop('Play', axis=1)
y = df['Play']

#Train
model = CategoricalNB()
model.fit(X, y)

test = pd.DataFrame({
    'Outlook': ['Sunny'],
    'Temperature': ['Cool'],
    'Humidity': ['High'],
    'Wind': ['Strong']
})

# Transform the test data using the fitted encoders
for col in test.columns:
    if col in encoders:
        test[col] = encoders[col].transform(test[col])

probabilities = model.predict_proba(test)
probabilities
```
