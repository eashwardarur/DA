```python
import pandas as pd
from sklearn.linear_model import Ridge
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
from sklearn.linear_model import LinearRegression,Ridge,Lasso
```

```python
df = pd.read_csv("https://github.com/Jeevita-R/ML_datasets/raw/refs/heads/main/House_price_prediction.csv")
df
```

```python
df_encoded = pd.get_dummies(df, drop_first=True,dtype=int) # drop_first is
```

```python
X=df_encoded.drop('Price',axis=1)
y=df_encoded.Price
```

```python
df_encoded.info()
```

```python
xtrain, xtest, ytrain, ytest = train_test_split(X, y, test_size=0.3,random_state=2)
```

```python
linear = LinearRegression()
```

```python
linear.fit(xtrain,ytrain)
```

```python
ypred=linear.predict(xtest)
```

```python
linear.score(xtrain,ytrain)
```
