1.Convert the following column using Label Encoding:

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
df = pd.DataFrame({'Color': ['Red', 'Blue', 'Green', 'Red', 'Green']})
df['Color'].unique()
le = LabelEncoder()
df['Color_Encoded'] = le.fit_transform(df['Color'])
df
```

2. Encode the following feature (Order: High School < Bachelor's < Master's < PhD):

```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder
df = pd.DataFrame({'Education': ['High School', "Bachelor's", "Master's", 'PhD', "Bachelor's"]})
oe = OrdinalEncoder(categories=[['High School', "Bachelor's", "Master's", 'PhD']])
df['Education_Encoded'] = oe.fit_transform(df[['Education']]).astype(int)
df
```

```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder, StandardScaler, MinMaxScaler
data = {
    'Gender':    ['Male', 'Female', 'Female', 'Male', 'Female'],
    'Education': ['High School', 'Bachelor', 'Master', 'PhD', 'Bachelor'],
    'Salary':    [25000, 40000, 50000, 80000, 45000]
}
df = pd.DataFrame(data)

# ── Q45a: One-Hot Encoding on Gender (drop_first avoids dummy variable trap) ──
df = pd.get_dummies(df, columns=['Gender'], dtype=int)
# Creates: Gender_Male (Female = reference category, Gender_Male=0)

# ── Q45b: Ordinal Encoding on Education ──
order = [['High School', 'Bachelor', 'Master', 'PhD']]
oe = OrdinalEncoder(categories=order)
df['Education_Encoded'] = oe.fit_transform(df[['Education']]).astype(int)

# ── Q46: StandardScaler on Salary ──
ss = StandardScaler()
df['Salary_Standard'] = ss.fit_transform(df[['Salary']])
print(f"Mean: {ss.mean_[0]:.2f} | Std: {ss.scale_[0]:.2f}")

# ── Q47: MinMaxScaler on Salary ──
mm = MinMaxScaler()
df['Salary_MinMax'] = mm.fit_transform(df[['Salary']])
print(f"Min: {mm.data_min_[0]} | Max: {mm.data_max_[0]}")

# ── Q48: Compare ──
print(df[['Salary', 'Salary_Standard', 'Salary_MinMax']].round(4))
```
